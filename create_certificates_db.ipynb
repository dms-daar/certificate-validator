```python
import pandas as pd
import os 
import random
import string
import shutil
```

```python

def get_rand_string(): 
    random_string = ''.join(random.choices(string.ascii_letters + string.digits, k=25))
    return (random_string)
```

```python
files = os.listdir("certificates")

df = pd.DataFrame({"path": files})
df["validation_code"] = [get_rand_string() for _ in range(len(df))]
df["validation_code"] = df["validation_code"] + ".pdf"

df
```

```python
for _, row in df.iterrows(): 

    path, val_code = row["path"], row["validation_code"]
    path = f"certificates/{path}"

    shutil.copy(path, f"verify/{val_code}")
```
